## **Introduction**

Salifort Motors, a leader in alternative energy vehicle manufacturing, has brought me on board as a newly hired data analytics professional. As part of their data team, I am tasked with analyzing employee turnover trends to support the Human Resources department in increasing retention and reducing the high cost of attrition.

In this capstone project, I will put into practice the full range of data analytics skills I’ve developed throughout the course. This includes setting up workflows, conducting exploratory data analysis (EDA), building predictive models, and communicating key insights to stakeholders.

Using Python, I will build and evaluate both statistical and machine learning models—including logistic regression, decision trees, and random forests—to predict whether an employee is likely to leave the company. ### **Company Background**

Salifort Motors is a fictional French-based company at the forefront of the alternative energy vehicle industry. With a global workforce of over 100,000 employees, Salifort is involved in every stage of the vehicle lifecycle—from research and design to production and distribution. Their innovative focus on electric, solar, algae, and hydrogen-powered vehicles has positioned them as a world leader in sustainable transportation.

Through vertical integration and a commitment to employee development, Salifort Motors aims to build a strong organizational culture. However, a recent rise in employee turnover has prompted leadership to take action. By leveraging data and predictive modeling, I will help the company identify key drivers of attrition and support efforts to foster a more engaged, satisfied, and stable workforce.

### **The Salifort Motors Scenario**

As part of Salifort Motors’ data analytics team, I have been tasked with addressing a growing concern: the high rate of employee turnover. Turnover at Salifort includes both voluntary resignations and involuntary terminations. Leadership is particularly concerned because of the high financial and operational costs associated with employee turnover (churn) trends. The company invests heavily in recruiting, onboarding, and upskilling its workforce—and losing employees means a loss of time, money, and talent.

To better understand this issue, Salifort's Human Resources department conducted an employee survey aimed at uncovering potential drivers behind the departures. Now, it’s my role as a data analytics professional to analyze that data and build a model that can help predict whether an employee is likely to leave. By identifying the key factors that contribute to turnover, the company hopes to improve employee satisfaction and retention, reduce hiring costs, and maintain a stable workforce.

### **Project Scope**

This project focuses on developing a predictive model that helps Salifort Motors proactively address employee turnover. The model will use variables such as job title, department, number of projects, average monthly hours, and other relevant features to determine the likelihood of an employee leaving the company.

The project involves the full data analytics lifecycle—starting with exploratory data analysis (EDA) to uncover patterns and trends, followed by the development and evaluation of both statistical and machine learning models. Specifically, I will build a logistic regression model and two tree-based machine learning models: decision tree and random forest. The final task will be to evaluate each model's performance and select a champion model that offers the best predictive accuracy, which can then be used to generate actionable business insights. These insights will ultimately be translated into a multi-page Power BI analytics report designed to communicate workforce trends, predictive modeling outcomes, and retention recommendations to stakeholders.

## **Dataset Structure**

This dataset, titled **HR_capstone_dataset.csv**, contains information on **15,000 employees** at a multinational vehicle manufacturing company. Each row represents a unique employee and captures their **attrition status** through the `left` variable. It includes **performance and workload indicators** such as `last_evaluation` (performance score), `number_project` (projects handled), and `average_monthly_hours` (average hours worked per month). **Tenure information** is provided via `time_spend_company` (years at the company). The dataset also includes **binary events**, such as `work_accident` and `promotion_last_5years`, and **categorical attributes** like `department` (functional area) and `salary` (low, medium, high), representing the employee's pay scale.

| Column Name             | Type   | Description                                                             |
|-------------------------|--------|-------------------------------------------------------------------------|
| satisfaction_level      | int64  | The employee’s self-reported satisfaction level [0–1]                   |
| last_evaluation         | int64  | Score of employee's last performance review [0–1]                       |
| number_project          | int64  | Number of projects employee contributes to                              |
| average_monthly_hours   | int64  | Average number of hours employee worked per month                       |
| time_spend_company      | int64  | How long the employee has been with the company (in years)              |
| work_accident           | int64  | Whether or not the employee experienced an accident while at work       |
| left                    | int64  | Whether or not the employee left the company                            |
| promotion_last_5years   | int64  | Whether or not the employee was promoted in the last 5 years            |
| department              | str    | The employee's department                                               |
| salary                  | str    | The employee's salary (low, medium, or high)                            |

This dataset offers a structured view of employee characteristics and outcomes, forming the foundation for analysis and predictive modeling in the employee turnover project.

#### **Determining Outliers in `tenure`**

When building predictive models, outliers can introduce noise, distort patterns, and reduce model performance. For example, in an employee retention setting, unusually long tenure periods might represent anomalies such as rehired retirees or legacy data inconsistencies. Alternatively, they may reflect legitimate but rare employee behavior. Identifying these outliers is crucial for ensuring robust model training and accurate predictions.

The primary objective of this project is to **predict employee turnover**, and preliminary analysis suggests that an employee's **tenure is likely to influence their departure decision**. While longer tenures often indicate job satisfaction and loyalty, unusually high values may reflect atypical employment patterns that do not generalize to most employees.

Although the `tenure` values in the dataset appear valid and plausible from a domain standpoint, it's still valuable to quantify how many observations could be considered outliers. Outlier definitions vary by use case, and **domain expertise is essential** for setting thresholds. A common approach for normally distributed variables is to use the **interquartile range (IQR)** and define outliers as values **1.5 × IQR above the third quartile (Q3)**.

However, the **distribution of `tenure` is right-skewed**, meaning a small portion of employees have stayed significantly longer than others. In such cases, using Q3 to calculate outlier thresholds can be misleading, as Q3 is already influenced by the long tail. This results in a high threshold that may overlook meaningful outliers.

To address this, I adopted a stricter criterion, modifying the threshold by adding 1.5 × IQR to the median instead of Q3. Since the median is not affected by extreme values, it provides a more stable central tendency for skewed data. This method generates a lower threshold for identifying outliers, which is better suited to capturing atypical tenure durations in right-skewed distributions.

#### **Key Difference Between Median and Q3**

- **Median (50th percentile)**: Represents the middle of the dataset and is unaffected by extreme values, making it ideal for skewed distributions.
- **Q3 (75th percentile)**: Sits closer to the tail in right-skewed data and may already be inflated by outliers, resulting in a less sensitive threshold.


#### **Handling Outliers in Skewed Employee Data**

In the context of right-skewed employment data:

- **Using Q3 + 1.5 × IQR** creates a high threshold, potentially failing to flag long-tenured employees as outliers.
- **Using Median + 1.5 × IQR** results in a lower threshold that better identifies those extreme cases without being influenced by the skew.

Since the histogram revealed a right-skewed distribution for the `tenure` variable, I used the **median + 1.5 × IQR** as the threshold for identifying outliers instead of the conventional **Q3 + 1.5 × IQR**. This approach captures a broader set of high values as outliers due to the longer tail on the right. It ensures that outlier detection remains aligned with the underlying data distribution and the goals of the turnover prediction model.

Whether or not to remove these outliers will depend on the sensitivity of the final model to such values. For example, **decision trees and random forests** are generally robust to outliers, whereas models like **logistic regression** may be more sensitive—making outlier handling an important factor in performance and interpretability.

```python
q1 = hr_df1['tenure'].quantile(0.25)
q3 = hr_df1['tenure'].quantile(0.75)
iqr = q3 - q1
median = hr_df1['tenure'].median()

lower_limit = q1 - 1.5 * iqr
upper_limit = median + 1.5 * iqr

print(f"Lower limit for outliers: {lower_limit}")
print(f"Upper limit for outliers: {upper_limit}")

outliers = hr_df1[(hr_df1['tenure'] < lower_limit) | (hr_df1['tenure'] > upper_limit)]

print(f"Number of outlier observations in 'tenure': {len(outliers)}")
```

Using the median instead of Q3—due to the right-skewed nature of the `tenure` data—resulted in a **stricter threshold** for outlier detection, which identified **1,886 outliers**, highlighting the long tail of higher tenure values in the dataset.

Next, I applied the **standard IQR method** (Q1 − 1.5 × IQR and Q3 + 1.5 × IQR) to compute the lower and upper bounds for detecting outliers. This method is less conservative and is typically used when the distribution is not heavily skewed.

```python
q1 = hr_df1['tenure'].quantile(0.25)
q3 = hr_df1['tenure'].quantile(0.75)
iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

print(f"Lower limit for outliers: {lower_limit}")
print(f"Upper limit for outliers: {upper_limit}")

outliers = hr_df1[(hr_df1['tenure'] < lower_limit) | (hr_df1['tenure'] > upper_limit)]

print(f"Number of outlier observations in 'tenure': {len(outliers)}")
```

Using the **standard IQR method** with Q1 and Q3 to define the outlier bounds for `tenure`, the calculated lower and upper limits were **1.5** and **5.5**, respectively. This approach identified **824 outlier observations**, which is fewer than the median-based method. This outcome reflects a **more lenient threshold**, as Q3 is already influenced by the right-skewed distribution, making it less sensitive to the long tail of higher tenure values.

Depending on the modeling technique used later, one may choose the more **conservative** or **standard** method of outlier detection. Certain models—such as **linear regression or logistic regression**—are more sensitive to outliers and may produce biased estimates if extreme values are present. In contrast, **tree-based models** like decision trees and random forests tend to be more robust to outliers.

Therefore, when reaching the modeling stage, it's important to **re-evaluate whether to retain or remove outliers** based on the **sensitivity of the chosen model**. Outlier handling should be a context-driven decision, guided by both the data distribution and the modeling framework.

## Milestone 4: Confirm model assumptions - Analyze and Construct

### Evaluating the Assumptions for Model Validity

**Logistic Regression model assumptions**
- Outcome variable is categorical
- Observations are independent of each other
- No severe multicollinearity among X variables
- No extreme outliers
- Linear relationship between each X variable and the logit of the outcome variable
- Sufficiently large sample size

Now, let's evaluate the assumptions for the Logistic Regression model's validity.

#### **Outcome variable is categorical**

The outcome variable, **`left`**, is categorical, representing employee attrition. A value of **0** indicates the employee stayed, while **1** indicates the employee left the company.

#### **Observations are independent of each other**

Yes, this assumption is reasonable. Each record represents an individual employee, and their decision to leave is presumed to be independent of others. Unless additional metadata (e.g., team structures or managerial relationships) suggests otherwise, it's appropriate to treat the observations as independent.

#### **No severe multicollinearity among X variables**

- **No severe multicollinearity among X variables**  
  There is no evidence of severe multicollinearity among the predictor variables, as seen in the correlation heatmap. Multicollinearity typically becomes a concern when variables exhibit high correlations (generally |correlation| ≥ 0.8 or 0.9). In this dataset, the highest observed correlations between predictors are:  
  - `number_project` & `average_monthly_hours`: **0.33**  
  - `last_evaluation` & `average_monthly_hours`: **0.26**  
  - `number_project` & `last_evaluation`: **0.27**  
  These values indicate only weak to moderate relationships and are well below thresholds that would raise multicollinearity concerns.

#### **No extreme outliers**

  While the HR dataset contains some mild outliers—such as slightly high values for `average_monthly_hours` or `number_project`—these appear plausible within a real-world context and do not violate key statistical assumptions. However, as noted earlier, employees with unusually long tenure stand out as statistical outliers. These long-tenured cases will be excluded from the analysis to avoid skewing results and to better meet model assumptions.

#### **Linear relationship between each X variable and the logit of the outcome variable**

```python
df = hr_df1.copy()
continuous_vars = ['satisfaction_level', 'tenure', 'average_monthly_hours',
                   'number_project', 'last_evaluation']
outcome = 'left'

plt.figure(figsize=(18, 10))

for i, var in enumerate(continuous_vars, 1):
    df_temp = df.copy()
    df_temp[f'{var}_bin'] = pd.qcut(df_temp[var], q=10, duplicates='drop')

    binned = df_temp.groupby(f'{var}_bin', observed=False).agg({
        var: 'mean',
        outcome: 'mean'
    })

    binned['logit'] = np.log((binned[outcome] + 1e-5) / (1 - binned[outcome] + 1e-5))

    plt.subplot(2, 3, i)
    plt.plot(binned[var], binned['logit'], marker='o')
    plt.xlabel(f'Mean {var} in Bin')
    plt.ylabel(f'Logit (log odds of {outcome})')
    plt.title(f'Logit Linearity: {var}')
    plt.grid(True)

plt.tight_layout()
plt.show()
```

To satisfy this assumption in logistic regression, each **continuous** predictor should have a **linear relationship** with the log-odds of the outcome. Here's what the plots suggest:

#### Continious Predictors

- **satisfaction_level**:  
  The relationship with the logit is **not linear**. The pattern shows a sharp drop at both extremes with a nonlinear curve in the middle, violating the assumption.

- **tenure**:  
  This variable **does not exhibit a linear or monotonic relationship** with the logit. The inverted V-shaped trend suggests a potential violation of the linearity of the logit assumption.

- **average_monthly_hours**:  
  The relationship is **curved and not linear**. 

- **number_project**:  
  The plot shows a **U-shaped** relationship with the logit, clearly **nonlinear**. 

- **last_evaluation**:  
  This variable shows a **strongly nonlinear (V-shaped)** pattern with the logit, violating the assumption. 

#### Categorical Predictors

The variables **work_accident**, **promotion_last_5years**, **department**, and **salary** were not checked for the linearity of the logit assumption because they are all categorical rather than continuous. For **binary categorical variables** like *work_accident* and *promotion_last_5years*, the model simply compares the log-odds between two groups, so the concept of a linear relationship with the logit does not apply. Similarly, **nominal variables** like *department* and **ordinal variables** like *salary* (even when encoded numerically) are treated as group comparisons rather than continuous trends. The linearity of the logit assumption is only relevant for **continuous predictors**, not for categorical variables encoded as dummy or ordinal values.

#### Observation

Only continuous numeric variables are subject to the **linearity of the logit** assumption; categorical variables — whether binary, nominal, or ordinal — are **excluded** from this check and are appropriately handled as group comparisons in logistic regression.

During the check for this assumption, **several continuous variables did not meet the linearity of the logit assumption**. While this can impact **coefficient interpretability** and potentially lead to **misestimated probabilities** in certain ranges, the overall **classification performance may still remain acceptable** — especially since the primary goal here is **prediction rather than inference**.

To better capture the non-linearity, I could consider applying transformations such as **quadratic terms, splines, or binning**, or even testing both linear and transformed versions for robustness. However, given the **moderate degree of non-linearity** observed, I’m comfortable proceeding without adjustment for now. This trade-off will be kept in mind for **future refinement**, particularly if the focus shifts toward **interpretation or inference**.

#### **Precision-Recall AUC (AUC-PR)**

Below is the Precision-Recall curve, which illustrates the trade-off between precision and recall at different thresholds, along with the calculated AUC-PR. I will write a function that will help me extract all the scores from the grid search. This function will allow me to retrieve and organize the performance metrics for each combination of hyperparameters tested during the search.

```python
def make_results(model_name:str, model_object, metric:str):
    '''
    Arguments:
        model_name (string): name to label the model in the output table  
        model_object: a fitted GridSearchCV object tuned with multiple scoring metrics  
        metric (string): one of 'precision', 'recall', 'f1', 'accuracy', or 'auc' to select the best model
    
    Returns a pandas DataFrame with F1, recall, precision, accuracy, and AUC scores  
    for the model that had the highest mean validation score for the chosen metric.
    '''


    # Create dictionary that maps input metric to actual metric name in GridSearchCV
    metric_dict = {'auc': 'mean_test_roc_auc',
                   'precision': 'mean_test_precision',
                   'recall': 'mean_test_recall',
                   'f1': 'mean_test_f1',
                   'accuracy': 'mean_test_accuracy'
                  }

    # Get all the results from the CV and put them in a df
    cv_results = pd.DataFrame(model_object.cv_results_)

    # Isolate the row of the df with the max(metric) score
    best_estimator_results = cv_results.iloc[cv_results[metric_dict[metric]].idxmax(), :]

    # Extract Accuracy, precision, recall, and f1 score from that row
    auc = best_estimator_results.mean_test_roc_auc
    f1 = best_estimator_results.mean_test_f1
    recall = best_estimator_results.mean_test_recall
    precision = best_estimator_results.mean_test_precision
    accuracy = best_estimator_results.mean_test_accuracy
  
    # Create table of results
    table = pd.DataFrame()
    table = pd.DataFrame({'model': [model_name],
                          'precision': [precision],
                          'recall': [recall],
                          'F1': [f1],
                          'accuracy': [accuracy],
                          'auc': [auc]
                        })
  
    return table
```

I use the function just defined to get all the scores from grid search.

All of these scores from the decision tree model are strong indicators of good model performance. 

Now, I define a function that gets all the scores from a model's predictions.

```python
def get_scores(model_name:str, model, X_test_data, y_test_data):
    '''
    Arguments:
        model_name (string): name to label the model in the output table  
        model: a fitted GridSearchCV object  
        X_test_data: numpy array of X_test data  
        y_test_data: numpy array of y_test data

    Returns:
        pandas DataFrame: A table containing precision, recall, F1, accuracy, and AUC scores for the model, 
        which is evaluated using the provided test data.
    '''


    preds = model.best_estimator_.predict(X_test_data)

    auc = roc_auc_score(y_test_data, preds)
    accuracy = accuracy_score(y_test_data, preds)
    precision = precision_score(y_test_data, preds)
    recall = recall_score(y_test_data, preds)
    f1 = f1_score(y_test_data, preds)

    table = pd.DataFrame({'model': [model_name],
                          'precision': [precision], 
                          'recall': [recall],
                          'f1': [f1],
                          'accuracy': [accuracy],
                          'AUC': [auc]
                         })
  
    return table
```

### Feature Engineering

I'm aware that the high evaluation scores may appear suspiciously optimistic, raising the possibility of data leakage. Data leakage occurs when information that should not be available during training — either because it belongs to the test set or reflects data that wouldn't be accessible in a real-world deployment — is inadvertently included in the training process. This can result in inflated performance metrics that fail to generalize and won’t hold up in production.

In my case, it’s likely that the company won't consistently have up-to-date satisfaction levels for all employees. Additionally, the `average_monthly_hours` feature might also be leaking information. ## Milestone 6: Evaluate model results - Analyze

#### **Key Evaluation Metrics for Classification Models**

- **AUC** is the area under the ROC curve; it's also considered the probability that the model ranks a random positive example more highly than a random negative example.
- **Precision** measures the proportion of data points predicted as True that are actually True, in other words, the proportion of positive predictions that are true positives.
- **Recall** measures the proportion of data points that are predicted as True, out of all the data points that are actually True. In other words, it measures the proportion of positives that are correctly classified.

Difference:
Precision evaluates how many of the instances predicted as positive by the model are actually true positives—it focuses on the accuracy of the model’s positive predictions. In contrast, recall takes a broader perspective by considering all actual positive cases in the dataset and measuring how many of them the model correctly identifies. While precision assesses the reliability of positive classifications, recall reflects the model’s ability to capture all true positives.

- **Accuracy** measures the proportion of data points that are correctly classified.
- **F1-score** is an aggregation of precision and recall.

#### **Evaluating Classification Models for Employee Attrition: A Comparative Analysis of Logistic Regression, Decision Trees, and Random Forests**

For this employee attrition analysis, I considered both **logistic regression** and **tree-based models**, including standalone **decision trees** and an ensemble model, **random forest**, across two rounds: one before addressing data leakage (Round 1) and one after (Round 2).

For both Decision Tree and Random Forest models (Rounds 1 and 2), I performed hyperparameter tuning using GridSearchCV with `refit='roc_auc'` to prioritize optimizing the ROC AUC metric. Alongside ROC AUC, I included key evaluation metrics—**accuracy**, **precision**, **recall**, and **F1 score**—to provide a well-rounded assessment of model performance and gain a comprehensive understanding of how the models predict employee turnover.

For the Decision Tree models, **cross-validation was set to 4-fold (cv=4)**, meaning each parameter combination was trained and validated **four times** using different train-test splits (3 folds for training, 1 for validation). This resulted in a total of **144 model fits** (36 combinations × 4 folds).

For the Random Forest models, the **cross-validation strategy was also set to 4-fold (cv=4)**, meaning each of the 108 parameter combinations was trained and validated four times using different data splits, resulting in **432 model fits**.

For each combination, the ROC AUC values from all four folds were averaged, and the parameter combination with the highest average was selected as the optimal configuration. This rigorous tuning process ensured the model was well-optimized, robust, and generalizable across different subsets of data, minimizing the risk of overfitting while maintaining strong performance.
